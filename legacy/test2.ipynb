```python
import re

DATE_RE   = re.compile(r'([A-Z][a-z]{2} \d{1,2}, \d{4})\s*·')
LINK_RE   = re.compile(r'https?://\S+')
AUTHOR_RE = re.compile(
    r'(?:by|author\s*=\s*[{(])\s*([\w\.\-\,\s&]+?)(?:[)}]|,|\.)',
    re.I
)
PREFIX_RE = re.compile(r'(title|link|snippet):', re.I)

def parse_snippets_to_markdown(raw: str) -> str:
    text = re.sub(r'\s+', ' ', raw.strip())
    anchors = [(m.start(), m.group(1)) for m in DATE_RE.finditer(text)]
    anchors.append((len(text), None))

    md_blocks = []

    for (start, date), (end, _) in zip(anchors[:-1], anchors[1:]):
        chunk = text[start:end].strip()

        # ---------- date (not printed, but kept if you want) ----------
        # date_str = date

        # ---------- link ----------
        link_match = LINK_RE.search(chunk)
        link = link_match.group(0) if link_match else ''

        # ---------- author ----------
        author_match = AUTHOR_RE.search(chunk)
        authors = author_match.group(1).strip() if author_match else ''

        # ---------- title ----------
        explicit_title = re.search(
            r'title:\s*(.*?)\s*(?=link:|snippet:|https?://|$)', chunk, re.I)
        if explicit_title:
            title = explicit_title.group(1).strip()
        else:
            pre_link = chunk.split(link)[0] if link else chunk
            cands = re.findall(r'([A-Z][\w\-\:\(\)\,\s]{15,})', pre_link)
            title = max(cands, key=len).strip() if cands else pre_link[:80].strip()

        # ---------- assemble markdown ----------
        block = f"# {title}\n## {authors}\n## {link}\n"
        md_blocks.append(block)

    return "\n".join(md_blocks)

# ------------- quick usage -------------
if __name__ == "__main__":
    RAW = """<PASTE THE WHOLE SNIPPET HERE>"""
    print(parse_snippets_to_markdown(results))
```
